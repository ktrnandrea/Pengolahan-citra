```python
planets = ['Mercury', 'Venus', 'Earth', 'Mars', 'Jupiter', 'Saturnus', 'Uranus', 'Neptunus']
for planet in planets:
  print(planet, end=' ')

print("\n")

for i in range(5):
  print("Doing important work. i =", i)

print("\n")

areas = [11.25, 18.0, 20.0, 10.75, 9.50]
for index, area in enumerate(areas):
  print("Room" + str(index) + ": " + str(area) + '°C')
```

```python
i = 0
while i < 0:
  print(i, end=' ')
  i += 1

print("\n")
while True:
  a = input('Input string: ')
  if a == 'quit':
    break
  print('Length of a is {}'.format(len(a)))

print("\n")

fruits = ['Apples', 'Banana', 'Orange', 'Grape', 'Pineaple']
for fruit in fruits:
  if fruit == 'Orange':
    continue
  print('My fruit is {}'.format(fruit))
```
